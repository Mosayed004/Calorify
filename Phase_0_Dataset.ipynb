```python
def get_image_links(query):
    url = f"https://www.bing.com/images/search?q={query}&form=HDRSC2&first=1&tsc=ImageBasicHover"

    headers = {
        "User-Agent": "Mozilla/5.0"
    }

    response = requests.get(url, headers=headers)
    soup = BeautifulSoup(response.text, "html.parser")

    links = []

    for a in soup.find_all("a", class_="iusc"):
        m = a.get("m")
        if m:
            try:
                m_json = json.loads(m)
                img_url = m_json.get("murl")
                if img_url:
                    links.append(img_url)
            except:
                pass

    return list(set(links))
```

```python
def download_images(links, folder, max_new_images=30):
    existing_files = len(os.listdir(folder))
    count = existing_files
    downloaded = 0

    for link in links:
        try:
            img_data = requests.get(link, timeout=8).content

            with open(f"{folder}/{count}.jpg", "wb") as f:
                f.write(img_data)

            count += 1
            downloaded += 1

            if downloaded >= max_new_images:
                break

            time.sleep(0.2)

        except:
            pass

    print(f"Downloaded {downloaded} new images to {folder}")
```

```python
def clean_folder(folder, min_size=(50, 50)):
    deleted = 0
    skipped = 0

    for file in os.listdir(folder):
        path = os.path.join(folder, file)

        try:
            with Image.open(path) as img:
                width, height = img.size

                if width < min_size[0] or height < min_size[1]:
                    os.remove(path)
                    deleted += 1

        except:
            skipped += 1

    print(f"Finished cleaning {folder} | Deleted {deleted} | Skipped {skipped}")
```

```python
for food in queries:
    clean_folder(f"dataset/{food}")
```

```python
import os
from PIL import Image
import imagehash

def remove_duplicates(folder, threshold=5):
    hashes = []
    files_seen = []
    deleted = 0

    for file in os.listdir(folder):
        path = os.path.join(folder, file)

        try:
            with Image.open(path) as img:
                img_hash = imagehash.phash(img)

            duplicate_found = False

            for i, h in enumerate(hashes):
                # لو الفرق صغير يبقى نفس الصورة
                if abs(img_hash - h) <= threshold:
                    duplicate_found = True
                    break

            if duplicate_found:
                os.remove(path)
                deleted += 1
            else:
                hashes.append(img_hash)
                files_seen.append(file)

        except:

            pass

    print(f"Finished duplicate removal in {folder} | Deleted {deleted} images")
```

```python
for food in queries:
    remove_duplicates(f"dataset/{food}")
```
